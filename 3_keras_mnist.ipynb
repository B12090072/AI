```python
# 從 tensorflow 核心庫中導入 keras。Keras 是一個高階神經網路 API，能讓我們用極其簡潔的程式碼建立深度學習模型。
from tensorflow import keras

# 從 keras 的 models 模組中導入 Sequential。
# Sequential 代表「線性堆疊模型」，它可以讓我們像樂高積木一樣，把各個網路層（Layers）由前到後、一層一層地依序堆疊起來。
from keras.models import Sequential

# 從 keras 的 layers 模組中導入 Dense（全連接層）與 Flatten（展平層）。
# Dense: 最基礎的全連接層，每一個輸入節點都會與輸出節點相連，用於學習特徵的組合。
# Flatten: 展平層。它的作用是把多維度的輸入資料（例如一張 28x28 像素的二維圖片）「壓扁」成一維的向量（變成 784 個一維數字），這樣後續才能送進 Dense 層進行處理。
from keras.layers import Dense, Flatten
```

```python
mnist=keras.datasets.mnist
(train_data, train_label), (test_data, test_label) = mnist.load_data()
```

```python
# 印出訓練特徵資料（train_data）的資料型態。
# 通常預期會顯示 <class 'numpy.ndarray'>，代表它是 NumPy 的多維陣列結構。
print(type(train_data))

# 印出訓練特徵資料（train_data）的矩陣形狀。
# 例如：如果顯示 (60000, 28, 28)，代表總共有 60000 筆資料，每筆資料是一張 28x28 像素的圖片。
print(train_data.shape)

# 印出訓練標籤資料（train_label，即正確答案）的資料型態。
# 同樣預期會是 NumPy 陣列。
print(type(train_label))

# 印出訓練標籤資料（train_label）的矩陣形狀。
# 例如：預期會顯示 (60000,) 或 (60000, 1)，必須跟 train_data 的總筆數（60000）完全對齊。
print(train_label.shape)

# 印出測試特徵資料（test_data，用來評估模型好壞、不參與訓練的資料）的矩陣形狀。
# 例如：通常會佔總資料的 10~20%，可能會顯示 (10000, 28, 28)。
print(test_data.shape)

# 印出測試標籤資料（test_label，測試集的正確答案）的矩陣形狀。
# 例如：預期會顯示 (10000,)，筆數必須與 test_data 完全一致。
print(test_label.shape)
```

```python
# 印出訓練集第一筆資料的「正確答案」（標籤）。
# 例如：在手寫辨識（MNIST）中，這可能會印出一個數字 5，代表第一張圖片寫的是數字五。
print(train_label[0])

# 印出訓練集第一筆資料的「原始特徵數據」（通常是一個多維矩陣）。
# 例如：如果是一張 28x28 的灰階圖片，這裡會印出一個包含 28 列、每列 28 個數字的二維陣列，
# 數字大小通常介於 0 到 255 之間（代表像素的黑白亮度）。
print(train_data[0])
```

```python
import matplotlib.pyplot as plt  # 導入繪圖庫 Matplotlib 的 pyplot 模組，用於將資料數據視覺化

# 使用 imshow() 函式將二維矩陣（train_data[0]，即第一張圖片的像素資料）轉換成圖像。
# cmap='binary' 代表採用「二進位/黑白」顯色模式：
# 在 Keras 的許多影像資料集中（如 MNIST），這會將數值大的地方顯示為黑色，數值小（0）的地方顯示為白色，非常適合用來觀察手寫字體。
plt.imshow(train_data[0], cmap='binary')

# 將繪製好的圖片視窗彈出並顯示在螢幕上。在視窗關閉之前，程式通常會暫停在這一行。
plt.show()
```

```python
# 初始化一個 Sequential 類型的模型對象，準備以線性堆疊的方式由前到後建立網路層
model = Sequential()

# 新增第一層為「Flatten（展平層）」：
# 它的功能是把輸入的多維矩陣（例如 28x28 像素的二維圖片）「壓扁」成一維向量（變為 784 個一維數字），好讓後續的全連接層可以讀取
model.add(Flatten())

# 新增第二層為「Dense（全連接隱藏層）」：
# 32 代表這層擁有 32 個神經元；
# activation='relu' 採用 ReLU 活化函數來引入非線性特徵，加速模型收斂；
# input_dim=784 代表輸入資料的特徵維度是 784（注意：因為前一層用了 Flatten，這裡的 input_dim 其實會被自動推導，通常可以省略）
model.add(Dense(32, activation='relu', input_dim=784))

# 新增第三層為「Dense（全連接輸出層）」：
# 10 代表輸出層有 10 個神經元（對應 0 到 9 共 10 個預測類別）；
# activation='sigmoid' 採用 Sigmoid 活化函數（註：多分類任務的輸出層通常更推薦使用 'softmax' 活化函數，讓 10 個輸出的機率總和等於 1）
model.add(Dense(10, activation='sigmoid'))

# 編譯模型，設定訓練時的各項核心參數：
# optimizer='rmsprop' 使用 RMSprop 優化器來動態調整學習率並更新權重；
# loss='sparse_categorical_crossentropy' 使用「稀疏多元交叉熵」作為損失函數。這非常適合用於「答案標籤（Label）為整數數字（如 0, 1, 2...9）」的多分類任務；
# metrics=['accuracy'] 設定模型在訓練與測試時的評估指標為「準確率（Accuracy）」
model.compile(optimizer='rmsprop', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(train_data,train_label,epochs=20,batch_size=512)

score=model.evaluate(test_data, test_label)
print(score[1])
```
