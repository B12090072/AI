```python
# 從 tensorflow 庫中導入 keras，這是目前最主流的高階深度學習框架
from tensorflow import keras

# 從 keras 的 models 模組中導入 Sequential（線性堆疊模型），用來依序建立網路層
from keras.models import Sequential

# 從 keras 的 layers 模組中導入 Dense，這代表「全連接層」（每一個輸入節點都與輸出節點相連）
from keras.layers import Dense

# 導入 numpy 庫並簡寫為 np，主要用於科學計算與矩陣、陣列的處理
import numpy as np

# 隨機產生 1000 筆特徵資料（Data），每筆資料包含 10 個介於 0 到 1 之間的 float 浮點數
data = np.random.random((1000, 10))

# 隨機產生 1000 筆標籤資料（Labels），數值上限為 2（不包含 2），也就是只有 0 或 1，用於二元分類
labels = np.random.randint(2, size=(1000, 1))

# 印出特徵資料 data 的矩陣形狀（Shape），預期會顯示 (1000, 10)
print(np.shape(data))

# 印出特徵資料中的第一筆（索引為 0），用來觀察這 10 個隨機生成的數字長怎樣
print(data[0])

# 印出標籤資料 labels 的矩陣形狀（Shape），預期會顯示 (1000, 1)
print(np.shape(labels))

# 印出標籤資料中的第一個答案（0 或 1），用來觀察第一筆資料對應的真實類別
print(labels[0])

# 初始化一個 Sequential 類型的模型對象，準備以線性堆疊的方式由前到後建立網路層
model = Sequential()

# 新增第一層全連接層（隱藏層）：
# 32 代表這層有 32 個神經元；
# activation='relu' 採用 ReLU 活化函數來引入非線性特徵；
# input_dim=10 代表輸入資料的特徵維度是 10（對應 data 的 10 個欄位）
model.add(Dense(32, activation='relu', input_dim=10))

# 新增第二層全連接層（輸出層）：
# 1 代表輸出層只有 1 個神經元（輸出一個機率值）；
# activation='sigmoid' 採用 Sigmoid 活化函數，將輸出值壓縮在 0 到 1 之間，非常適合二元分類
model.add(Dense(1, activation='sigmoid'))

# 編譯模型，設定訓練時的參數：
# optimizer='rmsprop' 使用 RMSprop 優化器來調整權重；
# loss='binary_crossentropy' 使用二元交叉熵作為損失函數（二元分類專用）；
# metrics=['accuracy'] 設定評估模型好壞的指標為「準確率（Accuracy）」
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy'])

# 開始訓練模型（Fit）：
# data 和 labels 是訓練用的輸入與標準答案；
# epochs=10 代表整套 1000 筆資料會完整輪流訓練 10 次；
# batch_size=32 代表每看 32 筆資料就更新一次神經網路的權重
model.fit(data, labels, epochs=10, batch_size=32)

# 評估模型效能，計算模型在該資料集上的損失值（Loss）與準確率（Accuracy）
score = model.evaluate(data, labels)

# 印出評估結果中的準確率（score[0] 是損失值，score[1] 是準確率）
print(score[1])
```
